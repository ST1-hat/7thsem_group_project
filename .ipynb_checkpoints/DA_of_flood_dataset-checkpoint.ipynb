```python
data['Flood Occurred'].value_counts() # how many 1 and 0 values in output
# 0 - no flood history , 1 - flood occured
```

```python
import matplotlib.pyplot as plt 

data['Flood Occurred'].value_counts().plot(kind='bar', width = 0.2, color ='g')
plt.show()
```

```python
# correlation analysis - how strongly two values are related
#Land Cover & soil type column are object type so they can't be used in correlation finding

numeric_data = data.select_dtypes(include='number')
c = numeric_data.corr()
c
```

```python
import seaborn as sns

plt.figure(figsize=(12,8))
sns.heatmap(c, annot=True, cmap='Blues')
plt.show()
```

```python
#which value occur most often and whether data is spread out

data.hist(figsize=(12,10), histtype='bar',rwidth=0.5)
```

```python
import seaborn as sns
# shows outlier and spread

plt.figure(figsize=(6,3))

sns.boxplot(
    x=data['Rainfall (mm)'],
    color='lightgreen',
    width=0.4
)

plt.title("Rainfall Distribution")
plt.show()
```

```python
plt.figure(figsize=(10,5))

sns.boxplot(data=data[['Rainfall (mm)',
                       'Temperature (°C)',
                       'Humidity (%)']], palette = 'cool')

plt.xticks(rotation=45)
plt.show()
```
